```python
import random
from pyspark.sql import SparkSession
from pyspark.sql.functions import col, lit
from pyspark.sql.types import StructType, StructField, StringType, IntegerType, LongType

spark = SparkSession.builder \
    .master("local[2]") \
    .appName("PySpark Fifty5Blue Interview") \
    .config("spark.sql.shuffle.partitions", "4") \
    .getOrCreate()

def prepare_data():
    # Schema for the data
    schema = StructType([
        StructField("id", StringType(), True),
        StructField("str", StringType(), True),
        StructField("int", LongType(), True),
        StructField("position", LongType(), True)
    ])

    # Generate data for DataFrame 1
    data_1 = []
    for i in range(1, 21):
        n = random.randint(
            a=random.randint(a=1000, b=1000),
            b=random.randint(a=10000, b=1000000000)
        )
        data_1.append((hex(n), str(n), n, i))

    # Generate data for DataFrame 2
    data_2 = []
    for i in range(1, 11):
        n = random.randint(
            a=random.randint(a=1000, b=10000000),
            b=random.randint(a=10000000, b=1000000000)
        )
        data_2.append((hex(n), str(n), n, i))

    # Create DataFrames
    data_frame_1 = spark.createDataFrame(data_1, schema=schema)
    data_frame_2 = spark.createDataFrame(data_2, schema=schema)

    return data_frame_1, data_frame_2
```

```python
def process_data(data_frame_1, data_frame_2):
    # 1. Display data_frame_1
    print("DataFrame 1:")
    data_frame_1.show(truncate=False)

    # 2. Display data_frame_2
    print("DataFrame 2:")
    data_frame_2.show(truncate=False)

    # 3. DataFrame 3: Display records where the id is less than 10 characters
    data_frame_3 = None
    # Placeholder for logic
    print("DataFrame 3:")
    if data_frame_3:
        data_frame_3.show(truncate=False)

    # 4. DataFrame 4: Perform an anti-join (left) on column "position" between DataFrame 1 and DataFrame 2
    data_frame_4 = None
    # Placeholder for logic
    print("DataFrame 4:")
    if data_frame_4:
        data_frame_4.show(truncate=False)



# Main function
def main():
    data_frame_1, data_frame_2 = prepare_data()
    process_data(data_frame_1=data_frame_1, data_frame_2=data_frame_2)

if __name__ == "__main__":
    main()
```
